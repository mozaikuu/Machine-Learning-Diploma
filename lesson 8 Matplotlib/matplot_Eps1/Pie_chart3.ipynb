```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = [70,20,10,50]
# plt.pie(data)
lab = ['Python', 'Javascript', 'HTML', 'Java']
plt.pie(data, labels=lab)

plt.title('A Pie Chart')
plt.show()
```

```python
data = [70,20,10,50]
lab = ['Python', 'Javascript', 'HTML', 'Java']
expl = [0,0,0,0]
co= ['yellow','red','green','#DD9804']
plt.pie(data, labels=lab,wedgeprops = {'edgecolor': 'blue'},colors=co,explode =expl,startangle=0,autopct="%1.2f%%",rotatelabels=True)

plt.title('A Pie Chart')
plt.show()
```
