```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline
plt.style.use('seaborn-v0_8')
```

```python
ages = [10, 30, 45, 30, 31, 46, 12, 16, 32, 23, 28, 29]

# ,edgecolor='black'

plt.hist(ages,edgecolor='black')

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
ages = [10, 30, 45, 30, 31, 46, 12, 16, 32, 23, 28, 29]
bi = [10,20,30,40,50,60]

me = np.mean(ages)
print(me)
plt.hist(ages, bins=bi ,edgecolor='black')

plt.axvline(me, color = 'red', linewidth = 2)


plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```
