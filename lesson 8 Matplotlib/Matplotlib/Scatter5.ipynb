```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline
plt.style.use('seaborn-v0_8')
```

```python
x_as= [5,7,8,5,6,7,9,2,3,4,4,4,2,6,3,6,8,6,4,1]
y_as= [7,4,3,9,1,3,2,5,2,4,8,7,1,6,4,9,7,7,5,1]

plt.scatter(x_as,y_as)
plt.title('First Scatter Plot')
plt.xlabel('x data')
plt.ylabel('y data')
plt.show()
```

```python
x= [5,7,8,5,6,7,9,2,3,4,4,4,2,6,3,6,8,6,4,1]
y= [7,4,3,9,1,3,2,5,2,4,8,7,1,6,4,9,7,7,5,1]

colors = [9,10,1,7,5,7,3,5,6,7,1,10,8,1,9,2,1,6,7,5]

plt.scatter(x,y,edgecolor = 'black',linewidth = 2,alpha = 0.5,c = colors)

plt.title('First Scatter Plot')
plt.xlabel('x data')
plt.ylabel('y data')
plt.show()
```

```python
x= [5,7,8,5,6,7,9,2,3,4,4,4,2,6,3,6,8,6,4,1]
y= [7,4,3,9,1,3,2,5,2,4,8,7,1,6,4,9,7,7,5,1]

colors = [9,10,1,7,5,7,3,5,6,7,1,10,8,1,9,2,1,6,7,5]
sizes = [209,486,381,255,191,315,185,228,174,538,239,394,399,153,273,293,436,501,397,539]

#color = 'red'
#edgecolor = 'black'
#linewidth = 2
#alpha = 0.5  ===> to light the color
#c = colors
#s = sizes
# cmap ='Greens'====>'https://matplotlib.org/stable/tutorials/colors/colormaps.html'


plt.scatter(x,y,edgecolor = 'black',linewidth = 1.5,alpha = 0.5,c = colors,s = sizes,cmap ='Blues')

# cbar = plt.colorbar()
# cbar.set_label('Colors bar')
plt.title('First Scatter Plot')
plt.xlabel('x data')
plt.ylabel('y data')
plt.show()
```
