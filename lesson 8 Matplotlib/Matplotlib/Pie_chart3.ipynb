```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = [70,20,10,50]
plt.pie(data)
lab = ['Python', 'Javascript', 'HTML', 'Java']
plt.pie(data, labels=lab)

plt.title('A Pie Chart')
plt.show()
```

```python
data = [70,20,10,50]
labels = ['Python', 'Javascript', 'HTML', 'Java']
explode = [0,0,0,0]
colors= ['yellow','red','green','#DD9804']
plt.pie(data, labels=labels,wedgeprops = {'edgecolor': 'black'},colors=colors,explode =explode,startangle=90,autopct="%1.1f%%",rotatelabels=True)

plt.title('A Pie Chart')
plt.show()
```
